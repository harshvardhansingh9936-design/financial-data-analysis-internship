```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Python environment is working!")
print("Pandas version:", pd.__version__)
```

```python
import yfinance as yf

ticker = "AAPL"

data = yf.download(
    ticker,
    start="2020-01-01",
    end="2025-01-01",
    auto_adjust=True
)

data.head()
```

```python
# Save AAPL data to the project's data folder
data.to_csv("../data/AAPL.csv")

print("AAPL data saved successfully!")
```

```python
# Load the saved AAPL data
aapl_loaded = pd.read_csv("../data/AAPL.csv", index_col=0, parse_dates=True)

# Display the first 5 rows
aapl_loaded.head()
```
